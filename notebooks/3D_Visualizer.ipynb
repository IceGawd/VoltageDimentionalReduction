```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import cm
from matplotlib.colors import Normalize
from mpl_toolkits.mplot3d import Axes3D
```

```python
# Read the CSV
dataset = "synthetic/torus"

df = pd.read_csv("../../Voltage_Data/" + dataset + ".csv")
```

```python
# Select dimension columns (start with 'd')
dims = [col for col in df.columns if col.startswith("d")]
x, y, z = df[dims[0]], df[dims[1]], df[dims[2]]

# Normalize labels for color mapping
labels = df["label"]
norm = Normalize(vmin=labels.min(), vmax=labels.max())
colors = cm.magma(norm(labels))

# Plot
fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111, projection="3d")
sc = ax.scatter(x, y, z, c=colors, s=50)

# Add colorbar
cbar = plt.colorbar(cm.ScalarMappable(norm=norm, cmap=cm.magma), ax=ax)
cbar.set_label("Label")

ax.set_xlabel(dims[0])
ax.set_ylabel(dims[1])
ax.set_zlabel(dims[2])
plt.savefig("../../Voltage_Temp/Scatter_Plots/" + dataset + "3D.png")
plt.show()
```
