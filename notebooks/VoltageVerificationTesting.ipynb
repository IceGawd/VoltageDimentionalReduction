```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
n = 11
points = np.array([[x] for x in np.arange(0,1,1.0/n)])
simple_set = setofpoints.SetOfPoints(points)
plt.plot(points)
```

```python
simple_problem = problem.Problem(simple_set, k=2, r=1)
simple_problem.getResistanceMatrix()
```

```python
def solve(r):
    simple_problem = problem.Problem(simple_set, r=r)
    simple_solver = solver.Solver(problem=simple_problem)
    return simple_solver.compute_voltages(landmark.Landmark(n//2, 1))
```

```python
print(solve(1))
plt.plot(solve(1))
```
