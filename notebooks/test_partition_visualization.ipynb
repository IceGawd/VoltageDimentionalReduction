# Testing Partition Visualization

This notebook contains tests for the partition visualization functionality of the Visualization class. We'll create various test cases to debug and verify the visualization behavior.

## 1. Generate Test Data Points

Create simple test data in 2D space:

```python
import sys
sys.path.append('../clean_code')  # Add path to import from clean_code directory

# Import our filter and visualization modules
from clean_code.filter import filter_by_voltage, partition_space
from clean_code.visualization import Visualization
```

```python
# Create test points in 2D space
np.random.seed(42)  # For reproducibility
n_points = 20
points = np.random.rand(n_points, 2) * 2 - 1  # Points in [-1, 1] range

# Create test landmarks/voltages
n_landmarks = 20
voltages = np.zeros((n_landmarks, n_points))

# Assign some random high voltages
for i in range(n_points):
    landmark_idx = np.random.randint(0, n_landmarks)
    voltages[landmark_idx, i] = np.random.rand() * 0.5 + 0.5  # Random values between 0.5 and 1.0

print(f"Generated {n_points} points and {n_landmarks} landmarks")
print(f"Points shape: {points.shape}")
print(f"Voltages shape: {voltages.shape}")
```

## 3. Test Voltage-based Filtering and Partitioning

Now let's test the filtering and partitioning functionality:

```python
# Test filtering
voltage_threshold = 0.7
filtered_indices = filter_by_voltage(voltages, voltage_threshold)
print(f"Filtered points with voltage threshold {voltage_threshold}:")
print(f"Kept {len(filtered_indices)} points out of {n_points}")

# Create partitions
partitions = partition_space(voltages, filtered_indices)
print("\nPartition information:")
print(f"Total partitions created: {len(partitions)}")
for i, partition in enumerate(partitions):
    print(f"Partition {i} contains {np.sum(partition)} points")
```

## 4. Test Partition Visualization

Now let's visualize the partitions using the Visualization class:

```python
# Visualize the partitions
print("Visualizing partitions...")
print(f"Points shape: {points.shape}")
print(f"Number of partitions: {len(partitions)}")
for i, partition in enumerate(partitions):
    print(f"Partition {i} shape: {partition.shape}, dtype: {partition.dtype}")

# Create the visualization
Visualization.plot_partitions(
    points=points,
    partitions=partitions,
    voltages=voltages
)
```
